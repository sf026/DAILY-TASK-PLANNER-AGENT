## DAILY TASK PLANNER AGENT

```python
from crewai import Agent, Task, Crew, LLM
```

### BLOG

```python
llm = LLM(
    model="groq/llama-3.1-8b-instant",
    api_key=" "
)
```

### AGENTS

```python
planner_agent = Agent(
    role="Productivity Planner",
    goal="Organize daily tasks efficiently and realistically",
    backstory="You are an expert in daily planning and time management.",
    llm=llm,
    allow_delegation=False,
    verbose=True
)
```

### TASKS

```python
daily_plan_task = Task(
    description=(
        "Create a simple and realistic daily plan for the following tasks:\n"
        "{tasks}\n\n"
        "Rules:\n"
        "- Organize tasks in a logical order\n"
        "- Assign approximate time slots\n"
        "- Keep the plan easy to follow\n"
        "- Output only the daily plan"
    ),
    expected_output="A clear daily schedule with time blocks and tasks.",
    agent=planner_agent
)
```

### DEFINE CREW

```python
daily_planner_crew = Crew(
    agents=[planner_agent],
    tasks=[daily_plan_task],
    verbose=True
)
```

### KICKOFF

```python
result = daily_planner_crew.kickoff(
    inputs={
        "tasks": "Gym, attend college classes, complete assignment, revise ML notes, sleep early"
    }
)
```
